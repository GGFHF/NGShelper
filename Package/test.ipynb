```python
def determine_genomic_zone(genomic_features_dict, seq_id, start, end):
    '''
    Determine the genomic zone where a variant is located.
    '''

    # initialize the genomic zone
    genomic_zone = ''

    # initialize control variables
    found_region = False
    found_gene = False
    found_mRNA = False
    found_exon = False
    found_CDS = False
    found_cDNA_match = False
    found_transcript = False
    found_tRNA = False
    found_pseudogene = False
    found_lnc_RNA = False
    found_snoRNA = False
    found_snRNA = False
    found_rRNA = False
    found_other = False

    # check found types
    for i in genomic_features_dict.keys():
        if genomic_features_dict[i]['type'] == 'region':
            found_region = True
        elif genomic_features_dict[i]['type'] == 'gene':
            found_gene = True
        elif genomic_features_dict[i]['type'] == 'mRNA':
            found_mRNA = True
        elif genomic_features_dict[i]['type'] == 'exon':
            found_exon = True
        elif genomic_features_dict[i]['type'] == 'CDS':
            found_CDS = True
        elif genomic_features_dict[i]['type'] == 'cDNA_match':
            found_cDNA_match = True
        elif genomic_features_dict[i]['type'] == 'transcript':
            found_transcript = True
        elif genomic_features_dict[i]['type'] == 'tRNA':
            found_tRNA = True
        elif genomic_features_dict[i]['type'] == 'pseudogene':
            found_pseudogene = True
        elif genomic_features_dict[i]['type'] == 'lnc_RNA':
            found_lnc_RNA = True
        elif genomic_features_dict[i]['type'] == 'snoRNA':
            found_snoRNA = True
        elif genomic_features_dict[i]['type'] == 'snRNA':
            found_snRNA = True
        elif genomic_features_dict[i]['type'] == 'rRNA':
            found_rRNA = True
        else:
            found_other = True

    print(f'found_region: {found_region}')
    print(f'found_gene: {found_gene}')
    print(f'found_mRNA: {found_mRNA}')
    print(f'found_exon: {found_exon}')
    print(f'found_CDS: {found_CDS}')
    print(f'found_cDNA_match: {found_cDNA_match}')
    print(f'found_transcript: {found_transcript}')
    print(f'found_tRNA: {found_tRNA}')
    print(f'found_pseudogene: {found_pseudogene}')
    print(f'found_lnc_RNA: {found_lnc_RNA}')
    print(f'found_snoRNA: {found_snoRNA}')
    print(f'found_snRNA: {found_snRNA}')
    print(f'found_rRNA: {found_rRNA}')
    print(f'found_other: {found_other}')

    # determine the genomic zone
    if not found_gene or not found_pseudogene:
        genomic_zone = 'intergenic'
    elif found_exon:
        genomic_zone = 'exonic'
    else:
        genomic_zone = 'intronic'

    # return the genomic zone where the variant is located
    return genomic_zone
```

```python
from goatools.obo_parser import GODag
from goatools.semantic import get_info_content, TermCounts, semantic_similarity

# Load the GO DAG (Directed Acyclic Graph)
godag = GODag('/home/fmm/Documents/Trabajo/PinusCanariensisXilogenesisGESU01.1-comparative/go.obo')

# Example GO terms
go_terms = ['GO:0008150', 'GO:0009987', 'GO:0050896']

# Fake gene-to-GO term mapping (to simulate term usage, required for info content calculation)
gene_to_go = {
    'gene1': ['GO:0008150', 'GO:0009987'],
    'gene2': ['GO:0050896', 'GO:0009987'],
    'gene3': ['GO:0009987']
}

# Create a TermCounts object for information content calculation
# -- termcounts = TermCounts(godag, gene_to_go.values())
termcounts = TermCounts(godag, gene_to_go)

# Calculate information content for each term
info_content = {go_id: get_info_content(go_id, termcounts) for go_id in go_terms}

# Calculate semantic similarity between pairs of GO terms
for i, go_id1 in enumerate(go_terms):
    for j, go_id2 in enumerate(go_terms):
        if i < j:
            sim = semantic_similarity(go_id1, go_id2, godag, info_content)
            print(f'Semantic similarity between {go_id1} and {go_id2}: {sim:.4f}')
```

```python
from goatools.obo_parser import GODag
from goatools.semantic import get_info_content, TermCounts, semantic_similarity
from collections import defaultdict

# cargar el archivo OBO
godag = GODag('/home/fmm/Documents/Trabajo/PinusCanariensisXilogenesisGESU01.1-comparative/go.obo')

# construir diccionario con los términos GO y sus respectivos genes
terms = ['GO:0008150', 'GO:0009987', 'GO:0050896']  # Ejemplo de términos GO
gene_to_go = {'gene1': ['GO:0008150', 'GO:0009987'], 'gene2': ['GO:0050896']}

# crear una instancia de TermCounts que contiene el conteo de términos GO
# -- termcounts = TermCounts(godag, gene_to_go.values())
termcounts = TermCounts(godag, gene_to_go)

# calcular la información de contenido para cada término GO
info_content = {go_id: get_info_content(go_id, termcounts) for go_id in terms}

# calcular la similitud semántica entre los términos (par a par)
similarities = defaultdict(dict)
for i, go_id1 in enumerate(terms):
    for j, go_id2 in enumerate(terms):
        if i < j:
            sim = semantic_similarity(go_id1, go_id2, godag, info_content)
            similarities[go_id1][go_id2] = sim
            similarities[go_id2][go_id1] = sim

# imprimir las similitudes calculadas
for go_id1 in similarities:
    for go_id2 in similarities[go_id1]:
        print(f'Similitud entre {go_id1} y {go_id2}: {similarities[go_id1][go_id2]}')
```
